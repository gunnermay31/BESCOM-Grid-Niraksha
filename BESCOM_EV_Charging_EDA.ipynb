### Part B — Infrastructure Location Planning
1. **Avoid top-25 outage-prone substations** for new EV charging stations
2. **BENGALURU zone** needs distributed charging across multiple feeders
3. **Prefer 220kV supply** over 66kV for reliability
4. **Weekend charging** offers ~10-15% more grid headroom
5. **Monsoon months** (Jun-Sep) require extra capacity buffer

### Part C — Key Grid Metrics
- Peak grid load: ~17,000 MW | Night valley: ~8,500 MW → **~8,500 MW spare at night**
- Average load factor: ~70% → significant room for smart scheduling
- Grid frequency: stable at 49.95-50.05 Hz (well within limits)
- YoY load growth: 5-15% → EV demand planning must account for organic growth
- 53,577 outage events/year → ~147/day average, concentrated 10AM-6PM
